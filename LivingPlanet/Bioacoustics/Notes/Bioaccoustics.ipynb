Tech allows us to *monitor scales that were not previously possible*. Cameras and accoustic automatic recorders allow scientitists to get a complete and integrative data over large time and geographic scales in a *cost-effective* way.

## Acoustic indices 

**Acoustic indices** examine the composition and diversity of all recorded vocalising organisms, as well as other acoustic features of the soundscape, such as environmental and human sounds. They are statistics that summarise some aspect of the distribution of acoustic energy and information in sound recordings. They *quantify the soundscape character* recorded at the site. 

Acoustic indices can be used to assess habitat quality in different locations or habitats, or to monitor changes in the environment over time. They make a good proxy for biodiversity or environmental quality, and have been linked with measures of species richness, community structure and phenology.

```r

library(tuneR) # For reading audio files 
library(soundecology) # For audio analysis functions 
library(seewave) # Another package for audio analysis functions 
library(dplyr) # To manipulation data 
library(ggplot2) # For plotting results 
library(tidyr) # For changing the shape of the datasets 
library(patchwork) # For combining plots
library(reshape2) #to change data from long to wide and visa versa
library(stringr) # For combining plots

# Set the path to your folder containing the audio files for one site
setwd("../Monkswood_dawn") 

# This creates a list of all the audio files (with .wav extension) in the specified folder 
file_list <- list.files(path =  "../Monkswood_dawn", pattern = "\\.wav$", full.names = TRUE) 

# Create empty lists to store the acoustic indices for each file 
# These lists will hold the results of the analysis for each file in the folder 
aci_scores <- numeric(length(file_list)) # For ACI scores 
ndsi_scores <- numeric(length(file_list)) # For NDSI scores 
bi_scores <- numeric(length(file_list)) # For BI scores 

# Loop through each file in the folder 
for (i in seq_along(file_list)) { 
  # Read the current audio files using the readWave function from tuneR package
  file <- readWave(file_list[i]) 
  
  # Calculate the Acoustic Complexity Index (ACI) 
  # calculates the ACI score for a the list of files and stores the result in a variable named aci_value
  aci_value <- acoustic_complexity(file, min_freq = 1000, max_freq = 11000)
  #store each calculated aci_value into a list called aci_scores
  aci_scores[i] <- aci_value 
  aci_scores
  
  # Calculate the NDSI 
  ndsi_value <- ndsi(file) 
  ndsi_scores[i] <- ndsi_value 
  ndsi_scores
  
  # Calculate the Bioacoustic index (Bi) 
  bi_value <- bioacoustic_index(file, min_freq = 1000, max_freq = 11000) 
  bi_scores[i] <- bi_value  
  bi_scores
}
```

**Acoustic Complexity Index (ACI)** determines the difference in amplitude between one time sample and the next within a frequency band, relative to the total amplitude within that band. Biophony is often of variable intensity, whilst antropophony such as engine noise is generally constant. Acoustically rich habitats may produce low ACI values if intensity does not vary greatly over time even if there are multiple contributing sound sources. 
High values might indicate storms, intermittent rain drops falling from vegetation, stridulating insects, or high levels of avian biophony.

**Bioacoustic index (BI)** is derived from the mean amplitude of individual frequency bands between 2 and 8 Hw minus that of the quiestest frequency band. 
High values are produced by recordings with high amplitude and greater disparity between loudest and quietest frequency bands.

**Normalised Difference Soundscape index (NDSI)** relies on the theoretical frequency split between anthropohony (1-2 Hz) and biophony (2-8 kHz). It is calculated from the power spectral density of the largest biophony band against that of the anthropophony band: 
$$ 
NDSI = \frac{Bio - Anthro}{Bio + Anthro} 
$$ 
It ranges from $[-1 , 1]$ with +1 indicating no sound in the anthropophony range.

$\Rightarrow$ *All these indicators reflect a rise in biophonic activity between 7 and 9 am, indicating effectively a dawn chorus in the area. The apparent correlation between NDSI and the two other graphs seem to indicate that most of the recorded sound is of biotic origin.*

#### Diversity across sites

```r
#Function to calculate Shannon Diversity Index 
shannon_diversity <- function(x) 
{ diversity(x, index = "shannon") 
}

# Calculate the diversity metrics for each site 
Monkswood_diversity <- Monkswood_subset %>% 
  summarize(Site = "Monkswood", 
            Species_Richness = n_distinct(common_name), #Finds the number of unique names 
            Shannon_Diversity = shannon_diversity(table(common_name)) 
            )

Parsonage_diversity <- Parsonage_subset %>% 
  summarize(Site = "Parsonage", 
            Species_Richness = n_distinct(common_name), #Finds the number of unique names 
            Shannon_Diversity = shannon_diversity(table(common_name)) 
  )

# Combine the results into a table 
diversity_table <- bind_rows(Monkswood_diversity, Parsonage_diversity) 
print(diversity_table)
```
