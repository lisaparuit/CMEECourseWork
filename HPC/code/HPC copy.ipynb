# Ecological Neutral Theory

## What is neutral theory? 
A theory in ecology doesn't carry the same weight as it does in physics for instance. It is not an explanation but a way of describing a general (or redundant) phenomena in Nature. 

ENT assumes that the demographic properties of all individuals are independant of its species/genetic identity. A model in which species are interchangeable is **neutral**. 

This theory was first designed for tropical forest canopy dynamics. Such ecosystem is, within the same trophic level (ie. the canopy), very diverse and rather govern by stochasticity than species-driven seletion forces. Indeed, in the canopy, individuals have a fixed identity who rarely die once established, and when they do, it is unlikely that the specific freed niche will be occupied by the best competitor because randomness prevades in this scenario.

## Ex: the voter model (Holley & Liggett 1975)
One can imagine that there is a "voter" at each point on a connected graph, where the connections indicate that there is some form of interaction between a pair of voters (nodes). The opinions of any given voter on some issue changes at random times under the influence of opinions of his neighbours. A voter's opinion at any given time can take one of two values, labelled 0 and 1. At random times, a random individual is selected and that voter's opinion is changed according to a stochastic rule. Specifically, one of the chosen voter's neighbors is chosen according to a given set of probabilities and that neighbor’s opinion is transferred to the chosen voter.

```python
import numpy as np
import matplotlib.pyplot as plt


def ListNeighbours(i1, j1):
    neighbours = []
    for i in range(-1, 2):
        for j in range(-1, 2):
            if (i == 0 and j == 0):
                continue
            ni = (i1 + i) % 20
            nj = (j1 + j) % 20
            neighbours.append((ni, nj))
    return neighbours

def OneIteration(grid):
    n= np.shape(grid)[0]
    for i in range(20):
        for j in range(20):
            if np.random.rand() < 0.5:
                neighbours = ListNeighbours(i, j)
                ni, nj = neighbours[np.random.choice(len(neighbours))]
                grid[i][j] = grid[ni][nj]

    return grid 

### Main program
np.random.seed(55)
fig, axs = plt.subplots(1, 5, figsize=(10, 5))

NbIterations = 10
grid = np.random.randint(2, size = (20,20))
# initial state
im0 = axs[0].imshow(grid, cmap='viridis', vmin=0, vmax=1)
axs[0].set_title('Initial State')
axs[0].axis('off')
# run iterations
for it in range(NbIterations):
    grid = OneIteration(grid)
# final state
im1 = axs[1].imshow(grid, cmap='viridis', vmin=0, vmax=1)
axs[1].set_title('10 iterations')
axs[1].axis('off')

NbIterations = 90
# run iterations
for it in range(NbIterations):
    grid = OneIteration(grid)
# final state
im1 = axs[2].imshow(grid, cmap='viridis', vmin=0, vmax=1)
axs[2].set_title('100 iterations')
axs[2].axis('off')

NbIterations = 150
# run iterations
for it in range(NbIterations):
    grid = OneIteration(grid)
# final state
im1 = axs[3].imshow(grid, cmap='viridis', vmin=0, vmax=1)
axs[3].set_title('250 iterations')
axs[3].axis('off')

NbIterations = 250
# run iterations
for it in range(NbIterations):
    grid = OneIteration(grid)
# final state
im1 = axs[4].imshow(grid, cmap='viridis', vmin=0, vmax=1)
axs[4].set_title('500 iterations')
axs[4].axis('off')


plt.tight_layout()
```
